### 5) Análisis de complejidad **[15 pts]**

El análisis considera el tamaño de entrada y las operaciones implementadas en el Notebook. Para el Max-Heap, \(n\) representa la cantidad de tareas. Para el grafo, \(V\) representa la cantidad de dispositivos y \(E\) la cantidad de conexiones posibles.

#### 5.1 Max-Heap

| Operación | Mejor caso | Peor caso |
|---|---:|---:|
| `MAX-HEAPIFY` | \(\Theta(1)\) | \(\Theta(\log n)\) |
| `BUILD-MAX-HEAP` | \(\Theta(n)\) | \(\Theta(n)\) |
| `EXTRACT-MAX` | \(\Theta(1)\) | \(\Theta(\log n)\) |
| Inserción en Max-Heap | \(\Theta(1)\) | \(\Theta(\log n)\) |

**Justificación:**

- `MAX-HEAPIFY`: en el mejor caso el nodo ya cumple la propiedad del Max-Heap y no requiere intercambios. En el peor caso el elemento desciende desde su posición hasta una hoja. La altura del Heap es \(O(\log n)\).
- `BUILD-MAX-HEAP`: recorre los nodos internos de abajo hacia arriba. Aunque una llamada individual a `MAX-HEAPIFY` puede costar \(O(\log n)\), la mayoría de los nodos están cerca de las hojas y descienden pocos niveles. Por eso el costo total es \(\Theta(n)\).
- `EXTRACT-MAX`: en el mejor caso el elemento que reemplaza la raíz ya cumple la propiedad. En el peor caso debe descender hasta una hoja, recorriendo la altura del Heap.
- Inserción: en el mejor caso el nuevo elemento permanece en la última posición. En el peor caso asciende desde una hoja hasta la raíz.

**Función de tiempo:** la altura de un Heap binario completo es proporcional a \(\log_2(n)\). Por ello, las operaciones de reparación, extracción e inserción tienen como cota de peor caso:

\$$[
T(n)=c_1+c_2\log_2(n)=\Theta(\log n)
\] $$

Para la construcción completa del Heap, el costo acumulado es:

\[$$
T(n)=\Theta(n)
\]$$

#### 5.2 Cálculo de distancias

| Operación | Mejor caso | Peor caso |
|---|---:|---:|
| Calcular una distancia | \(\Theta(1)\) | \(\Theta(1)\) |
| Generar todas las distancias | \(\Theta(n^2)\) | \(\Theta(n^2)\) |
| Dibujar los dispositivos | \(\Theta(n)\) | \(\Theta(n)\) |

**Justificación:** calcular una distancia requiere una cantidad constante de operaciones aritméticas. Al generar todas las conexiones posibles se recorren las parejas de dispositivos una sola vez. La cantidad de parejas es:

\[
T(n)=\frac{n(n-1)}{2}=\Theta(n^2)
\]

La cantidad de cálculos no depende de las coordenadas, por lo que el mejor y el peor caso son cuadráticos.

#### 5.3 Algoritmo de Kruskal

| Etapa | Mejor caso | Peor caso |
|---|---:|---:|
| Ordenar aristas con `sorted` | \(\Theta(E)\) | \(O(E\log E)\) |
| Operaciones Union-Find | \(O(\alpha(V))\) amortizado por operación | \(O(\alpha(V))\) amortizado por operación |
| Kruskal completo | \(\Theta(E)\) bajo condiciones favorables | \(O(E\log E)\) |

**Justificación:**

- El ordenamiento de Python puede aprovechar secuencias ya ordenadas, por lo que su mejor caso es lineal. En el peor caso ordena las \(E\) aristas en \(O(E\log E)\).
- Union-Find utiliza compresión de caminos y unión por rango. Sus operaciones tienen costo amortizado \(O(\alpha(V))\), donde \(\alpha\) es la función inversa de Ackermann.
- Kruskal recorre las aristas ordenadas y acepta las que unen componentes diferentes. El ordenamiento domina el peor caso: \(O(E\log E)\).

Para este problema el grafo es completo. El número de aristas es:

\[
E=\frac{V(V-1)}{2}
\]

Con seis dispositivos:

\[
E=\frac{6(6-1)}{2}=15
\]

Por tanto, la generación de distancias es \(\Theta(V^2)\) y el peor caso del proceso completo es \(O(V^2\log V)\). El espacio requerido es \(O(V+E)\), por el almacenamiento de dispositivos, aristas y estructuras auxiliares.

#### 5.4 Funciones T[n] y comparación de crecimiento

| Función | Interpretación |
|---|---|
| \(T(n)=1\) | Tiempo constante: \(O(1)\) |
| \(T(n)=\log_2 n\) | Crecimiento logarítmico: \(O(\log n)\) |
| \(T(n)=n\) | Crecimiento lineal: \(O(n)\) |
| \(T(n)=n\log_2 n\) | Crecimiento lineal-logarítmico: \(O(n\log n)\) |
| \(T(n)=n^2\) | Crecimiento cuadrático: \(O(n^2)\) |

## Punto 1. Cola de prioridad (Max-Heap)

**Teoría rápida.** Un Max-Heap es un árbol binario *casi completo* guardado en un arreglo. Con índices desde 0:

- hijo izquierdo de `i` → `2i + 1`
- hijo derecho de `i` → `2i + 2`
- padre de `i` → `(i - 1) // 2`

**Propiedad de Max-Heap:** cada nodo es mayor o igual que sus hijos, es decir `A[padre(i)] >= A[i]`. Así, la raíz (`A[0]`) siempre es el máximo.

- `MAX-HEAPIFY(A, i)`: asume que los subárboles de `i` ya son heaps y "hunde" `A[i]` hasta su lugar. Costo O(log n).
- `BUILD-MAX-HEAP(A)`: llama a `MAX-HEAPIFY` desde el último nodo interno (`n//2 - 1`) hasta la raíz. Costo O(n).

Cada tarea se guarda como tupla `(prioridad, código, descripción)` y se compara **solo por prioridad**.

````python
# Datos del enunciado: (codigo, tarea, prioridad)
tareas = [
    ("T01", "Configurar servidor", 6),
    ("T02", "Reparar sensor",      10),
    ("T03", "Actualizar software", 4),
    ("T04", "Revisar cámara",      8),
    ("T05", "Generar respaldo",    3),
    ("T06", "Configurar alarma",   7),
]

# Representación interna: (prioridad, codigo, tarea)  -> se compara por prioridad (índice 0)
arreglo_inicial = [(p, c, t) for c, t, p in tareas]

def fmt(heap):
    """Formato legible del arreglo: [T01:6, T02:10, ...]"""
    return "[" + ", ".join(f"{c}:{p}" for p, c, _ in heap) + "]"

# Funciones de índices
def izquierdo(i): return 2 * i + 1
def derecho(i):   return 2 * i + 2
def padre(i):     return (i - 1) // 2
````

````python
def max_heapify(A, i, heap_size):
    """MAX-HEAPIFY: hunde A[i] hasta que se cumpla la propiedad de Max-Heap en su subárbol."""
    l, r = izquierdo(i), derecho(i)
    mayor = i
    if l < heap_size and A[l][0] > A[mayor][0]:   # ¿el hijo izquierdo es mayor?
        mayor = l
    if r < heap_size and A[r][0] > A[mayor][0]:   # ¿el hijo derecho es aún mayor?
        mayor = r
    if mayor != i:                                # si algún hijo es mayor, intercambiar y seguir bajando
        A[i], A[mayor] = A[mayor], A[i]
        max_heapify(A, mayor, heap_size)


def build_max_heap(A, verbose=False):
    """BUILD-MAX-HEAP: convierte un arreglo cualquiera en Max-Heap (in place)."""
    n = len(A)
    for i in range(n // 2 - 1, -1, -1):           # de los nodos internos hacia la raíz
        max_heapify(A, i, n)
        if verbose:
            print(f"  max_heapify en i={i}: {fmt(A)}")


def es_max_heap(A):
    """Verificación: A[padre(i)] >= A[i] para todo i > 0."""
    return all(A[padre(i)][0] >= A[i][0] for i in range(1, len(A)))
````

````python
def dibujar_heap(A, titulo="Heap"):
    """Dibuja el heap como árbol binario (cada nodo: código y prioridad)."""
    n = len(A)
    niveles = int(math.log2(n)) + 1 if n else 1
    pos = {}
    for i in range(n):
        nivel = int(math.log2(i + 1))
        k = i - (2 ** nivel - 1)                       # posición dentro del nivel
        pos[i] = ((k + 0.5) / 2 ** nivel, -nivel)

    fig, ax = plt.subplots(figsize=(7, 1.2 + 1.3 * niveles))
    for i in range(1, n):                              # aristas padre-hijo
        x1, y1 = pos[padre(i)]; x2, y2 = pos[i]
        ax.plot([x1, x2], [y1, y2], color="gray", zorder=1)
    for i in range(n):                                 # nodos
        x, y = pos[i]
        p, c, _ = A[i]
        ax.scatter(x, y, s=1500, color="#cfe8ff" if i else "#ffd9a0", edgecolor="black", zorder=2)
        ax.text(x, y, f"{c}\n{p}", ha="center", va="center", fontsize=9, zorder=3)
        ax.text(x + 0.035, y + 0.28, f"i={i}", fontsize=7, color="gray")
    ax.set_title(titulo)
    ax.axis("off")
    ax.set_ylim(-niveles + 0.4, 0.6)
    ax.set_xlim(-0.08, 1.08)
    plt.show()


def mostrar_heap(A, titulo):
    print(f"{titulo}")
    print("  Arreglo:", fmt(A))
    print("  ¿Cumple propiedad Max-Heap?", es_max_heap(A))
    dibujar_heap(A, titulo)
````

````python
# 1) Arreglo inicial
heap = list(arreglo_inicial)           # copia para no modificar el original
print("ARREGLO INICIAL (sin ordenar como heap)")
print("  Arreglo:", fmt(heap))
print("  ¿Cumple propiedad Max-Heap?", es_max_heap(heap))
dibujar_heap(heap, "Árbol del arreglo inicial (todavía NO es Max-Heap)")

# 2) BUILD-MAX-HEAP (con pasos)
print("\nBUILD-MAX-HEAP paso a paso:")
build_max_heap(heap, verbose=True)

# 3) Max-Heap resultante y su árbol
print()
mostrar_heap(heap, "Max-Heap resultante")
````

**Lectura del resultado.** `BUILD-MAX-HEAP` empieza en `i = 2` (último nodo interno de 6 elementos): `T03(4)` se compara con su hijo `T06(7)` y se intercambian. Luego `i = 1`: `T02(10)` ya es mayor que `T04(8)` y `T05(3)`, no cambia. Por último `i = 0`: `T01(6)` se hunde hasta quedar bajo `T02(10)` y `T04(8)`. La raíz es la tarea de mayor prioridad: **T02 (10)**.

## Punto 2. Procesar tareas (EXTRACT-MAX e inserción)

**`HEAP-EXTRACT-MAX`**
1. Guardar la raíz `A[0]` (la tarea de mayor prioridad).
2. Mover el último elemento a la raíz y reducir el tamaño del heap.
3. Llamar `MAX-HEAPIFY(A, 0)` para restaurar la propiedad. Costo O(log n).

**`MAX-HEAP-INSERT`**: agregar al final y "subir" el elemento mientras sea mayor que su padre (*sift-up*). Costo O(log n).

````python
def extract_max(A):
    """EXTRACT-MAX: retira y devuelve la tarea de mayor prioridad, restaurando el heap."""
    if not A:
        raise IndexError("heap vacío")
    maximo = A[0]
    A[0] = A[-1]          # el último pasa a la raíz
    A.pop()               # el heap se reduce en 1
    if A:
        max_heapify(A, 0, len(A))
    return maximo


def insert_max_heap(A, tarea):
    """MAX-HEAP-INSERT: inserta al final y sube (sift-up) mientras sea mayor que su padre."""
    A.append(tarea)
    i = len(A) - 1
    while i > 0 and A[padre(i)][0] < A[i][0]:
        A[i], A[padre(i)] = A[padre(i)], A[i]
        i = padre(i)
````

````python
# Procesar las dos tareas de mayor prioridad
tareas_procesadas = []
for k in range(1, 3):
    p, codigo, descripcion = extract_max(heap)
    tareas_procesadas.append((p, codigo, descripcion))
    print(f"=== EXTRACT-MAX #{k} ===")
    print(f"Tarea procesada: {codigo} - {descripcion} - Prioridad {p}")
    mostrar_heap(heap, f"Heap después de la extracción #{k}")
    print()
````

````python
# Insertar T07 - Falla en sistema contra incendios - Prioridad 12
nueva = (12, "T07", "Falla en sistema contra incendios")
insert_max_heap(heap, nueva)
print("=== INSERCIÓN ===")
print("Se agregó: T07 - Falla en sistema contra incendios - Prioridad 12")
mostrar_heap(heap, "Heap después de insertar T07")
````

**Lectura del resultado.** `T07` (prioridad 12) entra como hoja y sube hasta la raíz porque es mayor que todos sus ancestros. Ahora sería la **próxima tarea** en atenderse, aunque haya llegado de última: eso es una cola de prioridad (no FIFO).

## Punto 3. Dispositivos a conectar

Cada dispositivo es un **vértice** con coordenadas `(x, y)`. Como cualquier par puede conectarse con cable, el grafo es **completo**: con 6 vértices hay `6·5/2 = 15` aristas. El peso de cada arista es la distancia euclidiana:

`distancia = sqrt((x2 - x1)^2 + (y2 - y1)^ 2)`

````python
# Datos del enunciado
dispositivos = {
    "A": ("Servidor", 2, 2),
    "B": ("Cámara",   7, 3),
    "C": ("Sensor",   4, 7),
    "D": ("Alarma",  10, 8),
    "E": ("Router",  13, 3),
    "F": ("Panel",   15, 7),
}

# Representación gráfica: posición de los seis dispositivos
fig, ax = plt.subplots(figsize=(7, 5))
for id_, (nombre, x, y) in dispositivos.items():
    ax.scatter(x, y, s=250, color="#4c9be8", edgecolor="black", zorder=3)
    ax.annotate(f"{id_} - {nombre}\n({x}, {y})", (x, y), textcoords="offset points",
                xytext=(0, 14), ha="center", fontsize=9)
ax.set_title("Posición de los dispositivos")
ax.set_xlabel("X"); ax.set_ylabel("Y")
ax.set_xlim(0, 17); ax.set_ylim(0, 10)
ax.set_aspect("equal"); ax.grid(alpha=0.3)
plt.show()
````

````python
def distancia(a, b):
    """Distancia euclidiana entre los dispositivos con id a y b."""
    _, x1, y1 = dispositivos[a]
    _, x2, y2 = dispositivos[b]
    return math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)


ids = list(dispositivos.keys())

# Lista de aristas (u, v, peso): una por cada pareja, sin repetir (u < v)
aristas = [(ids[i], ids[j], distancia(ids[i], ids[j]))
           for i in range(len(ids)) for j in range(i + 1, len(ids))]

print(f"Número de aristas: {len(aristas)}  (= 6·5/2)\n")
print("Distancia entre cada pareja de dispositivos:")
for u, v, w in aristas:
    print(f"  {u} - {v} : {w:.2f}")

# Matriz de distancias (vista complementaria)
print("\nMatriz de distancias:")
print("      " + "".join(f"{j:>7}" for j in ids))
for i in ids:
    print(f"  {i}   " + "".join(f"{distancia(i, j):>7.2f}" for j in ids))
````

## Punto 4. Árbol de Expansión Mínima (Kruskal manual)

Un **MST** conecta los `n` vértices con exactamente `n - 1` aristas (aquí **5**) y peso total mínimo, sin ciclos.

**Kruskal (estrategia voraz):**
1. Ordenar todas las aristas por peso de menor a mayor.
2. Recorrerlas en ese orden: aceptar la arista si conecta dos componentes **distintas**.
3. Evitar ciclos: se detectan con **Union-Find** (conjuntos disjuntos). Si `find(u) == find(v)`, ambos ya están conectados y la arista cerraría un ciclo.
4. Detenerse cuando se tengan `n - 1` aristas (los seis dispositivos conectados).

````python
class UnionFind:
    """Conjuntos disjuntos con compresión de caminos y unión por rango."""
    def __init__(self, elementos):
        self.padre = {e: e for e in elementos}
        self.rango = {e: 0 for e in elementos}

    def find(self, x):
        if self.padre[x] != x:
            self.padre[x] = self.find(self.padre[x])   # compresión de caminos
        return self.padre[x]

    def union(self, a, b):
        ra, rb = self.find(a), self.find(b)
        if ra == rb:
            return False                                # ya conectados -> habría ciclo
        if self.rango[ra] < self.rango[rb]:
            ra, rb = rb, ra
        self.padre[rb] = ra
        if self.rango[ra] == self.rango[rb]:
            self.rango[ra] += 1
        return True


def kruskal(vertices, aristas, verbose=True):
    """Devuelve (aristas_mst, costo_total)."""
    # 1. Ordenar aristas por peso
    ordenadas = sorted(aristas, key=lambda e: e[2])
    uf = UnionFind(vertices)
    mst, total = [], 0.0

    for u, v, w in ordenadas:
        if len(mst) == len(vertices) - 1:               # 4. detenerse con n-1 aristas
            break
        if uf.union(u, v):                              # 2-3. aceptar si no forma ciclo
            mst.append((u, v, w))
            total += w
            if verbose:
                print(f"  ACEPTA  {u} - {v} : {w:.2f}")
        elif verbose:
            print(f"  descarta {u} - {v} : {w:.2f}  (formaría un ciclo)")
    return mst, total


print("Aristas ordenadas por peso:")
for u, v, w in sorted(aristas, key=lambda e: e[2]):
    print(f"  {u} - {v} : {w:.2f}")

print("\nEjecución de Kruskal:")
mst, total_cable = kruskal(ids, aristas)
````

````python
print("Conexiones seleccionadas (MST):")
for u, v, w in mst:
    print(f"{u} - {v} : {w:.2f}")
print(f"Longitud total de cable: {total_cable:.2f}")

# Comprobaciones del enunciado
assert len(mst) == 5, "El MST debe tener exactamente 5 conexiones"
uf_check = UnionFind(ids)
assert all(uf_check.union(u, v) for u, v, _ in mst), "El MST no debe tener ciclos"
assert len({uf_check.find(i) for i in ids}) == 1, "Los 6 dispositivos deben quedar conectados"
print("\n✔ 5 conexiones, sin ciclos y los 6 dispositivos conectados.")
````

**Lectura del resultado.** Kruskal acepta `E-F`, `B-C`, `A-B` y `D-F`. Luego descarta `A-C` (ya están conectados vía `B`) y acepta `B-D`, que une las dos componentes `{A,B,C}` y `{D,E,F}`. Al llegar a 5 aristas se detiene.

*Nota sobre empates:* `A-B` y `D-F` pesan lo mismo (5.10), igual que `B-D` y `D-E` (5.83). El orden entre empates no cambia el costo total; `D-E` se descartaría igualmente por formar ciclo con `D-F` y `E-F`.
